## **Question 1: Define Data Transformation in ETL and explain why it is important.**

## Ans

Data Transformation is the second stage of the ETL (Extract, Transform, Load) process. It involves converting raw data extracted from different sources into a clean, consistent, and usable format before loading it into the target system.

Transformation may include activities such as removing duplicates, handling missing values, changing data types, standardizing formats, combining data, and creating new calculated columns.

It is important because raw data may contain errors, inconsistencies, missing values, or different formats. Data transformation improves the quality of data and makes it suitable for analysis, reporting, and decision-making.

Example: Converting different date formats like 10/09/2026 and 2026-09-10 into one common format such as 2026-09-10.

## **Question 2: List any four common activities involved in Data Cleaning.**

## Ans

Data Cleaning is the process of identifying and correcting problems in a dataset. Four common activities involved in Data Cleaning are:

1. Handling Missing Values: Missing data can be filled using suitable values such as mean, median, or mode, or the affected rows can be removed.

2. Removing Duplicates: Duplicate records are identified and removed to prevent the same information from being counted multiple times.

3. Correcting Data Types: Columns are converted into appropriate data types. For example, a date stored as text can be converted into a Date data type.

4. Standardizing Data: Different representations of the same information are converted into a common format. For example, "Male", "M", and "male" can all be standardized as "Male".

These activities improve the accuracy, consistency, and reliability of the dataset.

## **Question 3: What is the difference between Normalization and Standardization?**

## Ans

Normalization and Standardization are techniques used to transform numerical data into a comparable scale.

Normalization usually scales values into a fixed range, commonly between 0 and 1. It is generally performed using Min-Max Scaling.

Formula:
X' = (X - Min) / (Max - Min)

Standardization transforms the values based on their mean and standard deviation. Z-score Standardization generally produces data with a mean of 0 and a standard deviation of 1.

Formula:
Z = (X - Mean) / Standard Deviation

Therefore, Normalization uses a fixed range, while Standardization transforms data according to its mean and standard deviation.

## **Question 4: A dataset has missing values in the “Age” column. Suggest two techniques to handle this and explain when they should be used.**

## Ans

Two common techniques for handling missing values in the Age column are:

1. Mean/Median Imputation:
Missing Age values can be replaced with the mean or median age of the available records. Mean can be used when the data does not contain significant outliers. Median is generally better when outliers are present because extreme values have less effect on it.

Example: If the median age is 30, the missing Age values can be replaced with 30.

2. Removing Missing Records:
Rows containing missing Age values can be removed when the number of missing records is very small compared with the overall dataset.

This method should be avoided when many records have missing values because removing too many rows can result in the loss of useful information.

## **Question 6: What is One-Hot Encoding? Give an example with the categories: “Red, Blue, Green”.**

## Ans

One-Hot Encoding is a technique used to convert categorical data into numerical format. It creates a separate binary column for each category, where 1 represents the presence of a category and 0 represents its absence.

Example:

| Color | Red | Blue | Green |
|-------|-----|------|-------|
| Red   | 1   | 0    | 0     |
| Blue  | 0   | 1    | 0     |
| Green | 0   | 0    | 1     |

For example, if the color is Red, the Red column contains 1 while Blue and Green contain 0.

One-Hot Encoding is useful because many machine-learning algorithms require numerical input instead of text categories.

## **Question 7: Explain the difference between Data Integration and Data Mapping in ETL.**

## Ans

Data Integration is the process of combining data from multiple sources into a single unified dataset or system.

For example, a company may combine customer information from a MySQL database with sales information stored in Excel files.

Data Mapping, on the other hand, defines how fields or values from one source correspond to fields or values in another system.

Examples:
Cust_ID → Customer_ID
IN → India
US → United States

Therefore, Data Integration combines information from different sources, whereas Data Mapping defines the relationship between source and target data fields or values.

## **Question 8: Explain why Z-score Standardization is preferred over Min-Max Scaling when outliers exist.**

## Ans

Min-Max Scaling is highly sensitive to outliers because it uses the minimum and maximum values of the dataset to scale data into a fixed range, usually between 0 and 1.

For example, if most salaries are between ₹20,000 and ₹80,000 but one salary is ₹10,00,000, the extreme value becomes the maximum. As a result, most of the other salary values may get compressed into a very small range.

Z-score Standardization uses the mean and standard deviation instead of directly using the minimum and maximum values. Therefore, extreme minimum or maximum values do not compress the remaining data in the same way.

Hence, Z-score Standardization is generally more suitable than Min-Max Scaling when outliers are present. However, Z-score is also affected by extreme values because the mean and standard deviation can themselves be influenced by outliers.
