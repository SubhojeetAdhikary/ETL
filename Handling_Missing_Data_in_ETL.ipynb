## **Q1. What are the most common reasons for missing data in ETL pipelines?**

## Ans

Missing data in ETL pipelines can occur for several reasons:

1. Source System Issues: The original database or application may not have captured some values.

2. User Input Errors: Users may skip optional fields while filling forms or entering information.

3. Data Integration Issues: When data is combined from different sources, some fields may not exist in every source.

4. Extraction or Transfer Errors: Network failures, API issues, or interrupted data transfers may result in incomplete records.

5. System or Sensor Failure: Automated systems or sensors may fail to record data for a certain period.

6. Business Process Reasons: Some information may genuinely not be available or applicable for a particular customer or transaction.

Therefore, identifying why data is missing is important before deciding whether to delete, replace, or flag it.

## **Q2. Why is blindly deleting rows with missing values considered a bad practice in ETL?**

## Ans

Blindly deleting rows with missing values is considered a bad practice because it can result in the loss of useful information.

A row may contain one missing value but still contain valuable information in all its other columns. Deleting the complete row would unnecessarily remove that information.

It can also introduce bias if missing values are related to a particular type of customer, region, product, or business activity.

Therefore, before deleting rows, an ETL developer should first identify how much data is missing, understand why it is missing, and determine whether techniques such as imputation or flagging would be more appropriate.

## **Q3. Explain the difference between Listwise Deletion and Column Deletion. Also mention one scenario where each is appropriate.**

## Ans

Listwise Deletion:
Listwise deletion means removing the entire row if that row contains a missing value in an important field.

Example:
If only 1–2% of customer records have missing Region information and Region is essential for the analysis, those few rows may be removed.

Column Deletion:
Column deletion means removing an entire column when it contains too many missing values or when the column is not useful for analysis.

Example:
If a column contains 70–80% missing values and cannot be reliably recovered, deleting the complete column may be more appropriate.

Difference:
Listwise deletion removes records (rows), whereas column deletion removes a complete variable (column). The method should be selected according to the amount and importance of the missing data.

## **Q4. Why is median imputation preferred over mean imputation for skewed data such as income?**

## Ans

Median imputation is generally preferred over mean imputation for skewed data because the median is less affected by extreme values or outliers.

For example, suppose most customers have incomes between ₹40,000 and ₹80,000, but a few customers have incomes of ₹5,00,000 or more. These high values can significantly increase the mean.

The median represents the middle value of the dataset and is more resistant to such extreme observations.

Therefore, for skewed variables such as income, house prices, and transaction amounts, median imputation usually provides a more representative value than mean imputation.

## **Q5. What is forward fill and in what type of dataset is it most useful?**

## Ans

Forward Fill is a missing-value handling technique in which a missing value is replaced with the most recent previous non-missing value.

Example:

Before:

Date 1 = 100

Date 2 = NaN

Date 3 = 120

After Forward Fill:

Date 1 = 100

Date 2 = 100

Date 3 = 120

Forward fill is most useful for time-series or sequential data where the previous observation can reasonably represent the missing next observation.

Examples include stock prices, sensor readings, monthly sales records, and other time-based datasets.

## **Q6. Why should flagging missing values be done before imputation in an ETL workflow?**

## Ans

Flagging missing values should be done before imputation because once a missing value is replaced, it becomes difficult to identify which values were originally missing.

A separate flag column can be created before imputation.

For example:

Income = NaN → Income_Missing_Flag = 1
Income = 65000 → Income_Missing_Flag = 0

After this, the missing Income values can be imputed.

This approach preserves information about the original missingness and helps analysts distinguish between actual values and values that were filled during data transformation.

## **Q7. Consider a scenario where income is missing for many customers. How can this missingness itself provide business insights?**

## Ans

Missing income values can themselves provide useful business information rather than simply being treated as errors.

For example, if income is frequently missing for customers from a particular region, application channel, or customer group, it may indicate that income information is not being collected properly in that area.

It may also indicate that certain customers are less willing to disclose their income or that a particular system or form allows users to skip the field.

By creating a missing-value flag and analysing it with other variables, a business can identify patterns in missing data and improve its data collection process.

Therefore, missingness itself can sometimes reveal customer behaviour, operational problems, or weaknesses in the data collection system.

## **Q9. Imputation — Handle missing values in Monthly_Sales using Forward Fill**

## Ans

### Show Monthly_Sales before forward fill
    print("Monthly_Sales before Forward Fill:")
    display(df[['Customer_ID', 'Name', 'Monthly_Sales']])

### Apply forward fill
    df_ffill = df.copy()
    df_ffill['Monthly_Sales'] = df_ffill['Monthly_Sales'].ffill()

### Show Monthly_Sales after forward fill
    print("Monthly_Sales after Forward Fill:")
    display(df_ffill[['Customer_ID', 'Name', 'Monthly_Sales']])

##Explanation:
Forward Fill replaces each missing Monthly_Sales value with the previous available non-missing value.

Therefore:
Anjali Rao: NaN → 12000
Neha Singh: NaN → 15000
Karan Shah: NaN → 18000

Forward Fill can be suitable when the records have a meaningful sequence and the previous observation is considered a reasonable estimate for the missing value.

However, in a real business dataset, forward fill should only be used when the ordering of records has a logical meaning, such as time-series data.
