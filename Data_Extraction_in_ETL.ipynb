## **Question 1: Describe different types of data sources used in ETL with suitable examples.**

## Ans

In ETL (Extract, Transform, Load), data can be extracted from different types of sources depending on where an organization stores its information. Common data sources include:

1. Relational Databases: Data is stored in tables containing rows and columns. SQL is generally used to extract the required data.
Examples: MySQL, PostgreSQL, Oracle Database, SQL Server.
2. CSV and Text Files: CSV files store data in plain-text tabular format. They are commonly used for transferring data between different systems.
Example: A sales.csv file containing Product ID, Product Name, Quantity and Sales Amount.
3. Excel Files: Excel files are commonly used in businesses for maintaining reports, financial information and small datasets. ETL tools can extract data from specific sheets or tables.
Example: An Excel workbook containing monthly sales reports.
4. APIs: APIs allow applications to communicate and exchange data. ETL systems can use APIs to collect data from websites and online services.
Example: Extracting weather information from a weather API.
5. Cloud Storage: Modern organizations also store data in cloud platforms, from where ETL systems can extract it.
Examples: Amazon S3, Google Cloud Storage and Azure Blob Storage.

Therefore, ETL systems can work with structured as well as semi-structured data coming from multiple sources.

## **Question 2: What is data extraction? Explain its role in the ETL pipeline.**

## Ans

**Data extraction** is the process of collecting or retrieving data from one or more source systems so that it can be processed further.

It is the first stage of the ETL process, where ETL stands for Extract, Transform and Load.

During extraction, data can be collected from databases, CSV files, Excel files, APIs, cloud storage and other applications.

For example, suppose a company stores customer information in a MySQL database and sales information in Excel files. During extraction, the required information is collected from both sources.

After extraction, the collected data is sent to the transformation stage, where it can be cleaned, standardized and prepared before being loaded into the destination system such as a data warehouse.

## **Question 3: Explain the difference between CSV and Excel in terms of extraction and ETL usage.**

## Ans

CSV and Excel are both commonly used as data sources, but they have some important differences.

| Feature          | CSV                                  | Excel                                            |
| ---------------- | ------------------------------------ | ------------------------------------------------ |
| File Format      | Plain text                           | Spreadsheet format                               |
| Extension        | `.csv`                               | `.xlsx` or `.xls`                                |
| Multiple Sheets  | Not supported                        | Supported                                        |
| Formatting       | Does not store formatting            | Supports formatting                              |
| Formulas         | Not supported                        | Supported                                        |
| File Size        | Usually smaller                      | Usually larger                                   |
| Extraction Speed | Generally faster                     | Can be slower                                    |
| ETL Usage        | Better for large/simple tabular data | Better for business reports and smaller datasets |

CSV files are generally easier and faster to process because they contain simple text-based data. Excel files can contain multiple worksheets, formulas, formatting and other features, which can make extraction more complex.

For ETL processes involving large amounts of simple tabular data, CSV is usually preferred, while Excel is useful when business users maintain data in spreadsheets.

## **Question 4: Explain the steps involved in extracting data from a relational database.**



## Ans

The main steps involved in extracting data from a relational database are:

1. Establish Database Connection: Connect to the database using the required server name, database name, username and password.

2. Identify Required Tables: Determine which tables and columns contain the required information.

3. Write SQL Queries: SQL queries such as SELECT are used to retrieve the required records.
Example:
SELECT Customer_ID, Name, Sales FROM Customers;

4. Apply Filters: Conditions can be applied using WHERE clauses to extract only necessary data instead of the entire database.
5. Execute the Query: Run the SQL query against the database to retrieve the required records.
6. Validate Extracted Data: Check whether the extracted data is complete, accurate and contains the expected number of records.
7. Store or Transfer Data: Finally, the extracted data is transferred to the transformation stage or temporarily stored in a staging area.

These steps help ensure that only relevant and accurate data is extracted from the database.

## **Question 5: Explain three common challenges faced during data extraction.**

##  Ans

Three common challenges faced during data extraction are:

1. Poor Data Quality:
Source data may contain missing values, duplicate records, incorrect information or inconsistent formats. Such problems need to be identified and handled during the ETL process.

2. Different Data Formats:
Data may come from different sources such as databases, CSV files, Excel files and APIs. Each source may store information differently, making extraction and integration more difficult.

3. Large Volume of Data:
Extracting millions of records can require significant processing time, memory and network resources. ETL developers may need to extract data in smaller batches or use incremental extraction techniques.

Therefore, proper planning and validation are important for efficient data extraction.

## **Question 6: What are APIs? Explain how APIs help in real-time data extraction.**


## Ans

**API** stands for **Application Programming Interface**. It is a mechanism that allows two software applications to communicate and exchange information with each other.

In ETL, APIs can be used to extract data directly from websites, applications and online services.

For example, a weather application can use a weather API to retrieve the latest temperature, humidity and weather conditions.

APIs are useful for real-time or near-real-time data extraction because an ETL system can send requests to an API at regular intervals and receive updated information.

Many APIs provide data in formats such as JSON or XML, which can then be transformed and loaded into a database or data warehouse.

## **Question 7: Why are databases preferred for enterprise-level data extraction?**

## Ans

Databases are preferred for enterprise-level data extraction because they can efficiently store, manage and retrieve large amounts of structured data.

Databases provide several advantages:

- High performance: They can efficiently process millions of records.
- SQL support: SQL makes it easier to filter, join, aggregate and retrieve required information.
- Data consistency: Databases use rules and constraints to maintain accurate and consistent data.
- Security: Access can be controlled using usernames, passwords, roles and permissions.
- Scalability: Databases can handle increasing amounts of organizational data.
- Concurrent access: Multiple users and applications can access the database at the same time.

For these reasons, databases such as MySQL, PostgreSQL, Oracle and SQL Server are widely used as data sources in enterprise ETL systems.

## **Question 8: What steps should an ETL developer take when extracting data from large CSV files (1GB+)?**

## Ans

When working with a large CSV file of 1GB or more, loading the complete file into memory at once may cause performance or memory problems. An ETL developer should follow these steps:

1. **Inspect the File**: Check the columns, data types, delimiter, encoding and overall structure of the CSV file.
2. **Read Data in Chunks**: Instead of loading the entire file into memory, divide it into smaller chunks and process them one at a time.
3. **Select Required Columns**: Extract only the columns required for analysis or processing instead of reading unnecessary data.
4. **Apply Proper Data Types**: Assign suitable data types to columns to reduce memory consumption.
5. **Filter Unnecessary Records**: If possible, remove unwanted rows during extraction rather than processing them later.
6. **Process Data in Batches**: Transform and load each batch separately so that system memory is not overloaded.
7. **Validate the Data**: Check for missing values, duplicate records, incorrect data types and corrupted rows.
8. **Monitor Performance**: Monitor memory usage, processing time and errors while processing the large file.

For example, in Python, Pandas read_csv() with the chunksize parameter can be used to process a large CSV file in smaller batches.

This approach makes the ETL process more memory-efficient, reliable and scalable when dealing with very large files.
